```python
b=np.random.randint(9,18,(3,3))
print(b)
```

```python
# flatten --- used to convert any dimentiom of array into 1D array, a copy

flat=b.flatten()
print(flat)
```

```python
# ravel --- used to convert any dimentiom of array into 1D array, a reference

ral=b.ravel()
print(ral)
```

```python
# diffrence between flatten and ravel
# flatten creates a copy of the array, and changes made to it will not affect the original data
# ravel does not create a copy rather it only referes to the original data, and changes made to ravel will change the original data aswell

flat=b.flatten()
flat[1]=1234
print(flat)

ral=b.ravel()
ral[1]=1234
print(b)
```

```python
arr=np.random.randint(0,101,100).reshape(10,10)
print(arr)
min=0
max=0
for i in np.nditer(arr):
    if i>=max:
        max=i
    elif i<=min:
        min=i
print(f"max value in the matrix {max}")
print(f"min value in the matrix {min}")
```

```python
# arithmetic operations on 1D matrix
# + - / *
a=np.array([1,2,4,5,6])
b=np.array([4,2,5,8,2])

print(a+b)
print(a-b)
print(a/b)
print(a*b)
```

```python
# arithmetic operations on 2D matrix

c=np.array([[1,2],[4,2]])
d=np.array([[4,2],[1,2]])

print(c)
print()
print(d)
print()

print(c+d)
print(c-d)
print(c/d)
print(c*d) #it does not give the real matrix multiplication
print(c@d) # @, np.dot(a,b) and np.matmul(a,b) are used to do matrix multiplication
```

```python
# stacking, merge, concatinate
e=np.array([[1,2],[4,2]])
f=np.array([[4,2],[1,2]])
print(e)
print()
print(f)
print()
```

```python
# horizontal stacking
print(np.hstack((e,f)))
```

```python
# vertical stacking
print(np.vstack((e,f)))
```
