```python
import time
import numpy as np
size=10000000
# list
a=list(range(size))

start_a=time.time()
a=list(map(lambda i: i**2,a))
end_a=time.time()-start_a

# array
a=np.arange(size)
start_b=time.time()
a=[a**2]
end_b=time.time()-start_b

print("time taken by list",end_a)
print("time taken by array",end_b)
```

```python
# 1D array usng function, 1 to 100

arr=np.array([np.arange(1,100)])
print(arr)
```

```python
# 1D array using finction, 1-50 with steps of 5
arr=np.array([np.arange(0,50,5)])
print(arr)
```

```python
# 1D array of first 10 even numbers
arr=np.array(np.arange(2,10,2))
print(arr)
```

```python
# creating 2D array using numpy arange()
arr=np.arange(1,10)
print(arr)
```

```python
# .reshape
# used to reshape an array into a matrix 
arr=np.arange(1,10).reshape(3,3)
print(arr)
```

```python
# create a matrix of 9x9 
arr=np.arange(1,82).reshape(9,9)
print(arr)
```

```python
# .random
# prints float random values between 0-1
n=np.random.random(20)
print(n)
# it can also be reshaped
print(n.reshape(4,5))
```

```python
# randint
# returns random integer(s) between two integers
# random.randint(from,to,how many numbers)
n=np.random.randint(12)
print(n)
```

```python
# got 6 random integers and reshaped them into 2x3 matrix
n=np.random.randint(10,70,6).reshape(3,2)
print(n)
```

```python
# create an array, of random integer numbers from 1 to 51 and reshape the array into the possible formats
n=np.random.randint(1,51,36)
print(n.reshape(6,6))

print(n.reshape(2,18))
print(n.reshape(18,2))
print(n.reshape(9,4))
print(n.reshape(4,9))
print(n.reshape(12,3))
print(n.reshape(3,12))
print(n.reshape(1,36))
print(n.reshape(36,1))
```

```python
# zeros
# returns desired amount of zeros as float datatype
a=np.zeros(4)
print(a)

# it allows us to make matrix without the use of .reshape
a=np.zeros([2,2])
print(a)
```

```python
# 5x5 matrix of zeros witht eh help of reshape and without using reshape
# using reshape
a=np.zeros(25).reshape(5,5)
print(a)
# wihtout the use oof reshape

a=np.zeros([5,5])
print(a)
```

```python
# ones
# returns desired amount of ones as float datatype
a=np.ones(4)
print(a)

# it allows us to make matrix without the use of .reshape
a=np.ones([2,2])
print(a)
```
