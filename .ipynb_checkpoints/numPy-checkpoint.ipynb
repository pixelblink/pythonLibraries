```python
import time
import numpy as np
size=10000000
# list
a=list(range(size))

start_a=time.time()
a=list(map(lambda i: i**2,a))
end_a=time.time()-start_a

# array
a=np.arange(size)
start_b=time.time()
a=[a**2]
end_b=time.time()-start_b

print("time taken by list",end_a)
print("time taken by array",end_b)
```

```python
# linspace() --- linear space array, which means it creates an arrray where each elements will have same size diffrence
a=np.linspace(0,20,5)
print(a)
```

```python
b=np.random.randint(9,18,(3,3))
print(b)
```

```python
# flatten --- used to convert any dimentiom of array into 1D array, a copy

flat=b.flatten()
print(flat)
```

```python
# ravel --- used to convert any dimentiom of array into 1D array, a reference

ral=b.ravel()
print(ral)
```

```python
# diffrence between flatten and ravel
# flatten creates a copy of the array, and changes made to it will not affect the original data
# ravel does not create a copy rather it only referes to the original data, and changes made to ravel will change the original data aswell

flat=b.flatten()
flat[1]=1234
print(flat)

ral=b.ravel()
ral[1]=1234
print(b)
```
