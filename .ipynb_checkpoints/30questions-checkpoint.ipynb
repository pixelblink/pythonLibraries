```python
# 1. Make an array with the numbers 1 to 10.
print(np.arange(1,11))
```

```python
# 2. Print the first number and the last number of the array [4, 8, 15, 16, 23, 42].
a=[4, 8, 15, 16, 23, 42]
print(a[0])
print(a[-1])
```

```python
# 3. Predict the output: a = np.array([1, 2, 3]) then print(a * 2)
a = np.array([1, 2, 3]) # output will be [1 2 3]
print(a * 2) # output will be [2 4 6]
```

```python
# 4. Make an array of 5 zeros.
print(np.zeros(5,dtype=int))
```

```python
# 5.   Find the sum, the average (mean) and the biggest number of [3, 7, 2, 9, 5].
a=np.array([3, 7, 2, 9, 5])
print(sum(a))
print(np.mean(a))
print(max(a))
```

```python
# 6.   Print only the numbers bigger than 50 from [12, 67, 45, 89, 23, 71].
a=np.array([12, 67, 45, 89, 23, 71])
print(a[a>50])
```

```python
# 7.   Make a 3 x 3 array where every number is 1.
np.ones(9,dtype=int).reshape(3,3)
```

```python
# 8.   From [10, 20, 30, 40, 50, 60], print the first 3 numbers.
a=np.array([10, 20, 30, 40, 50, 60])
print(a[[0,1,2]])
```

```python
# 9.   Add 5 to every number in the array [1, 2, 3, 4].
a=np.array([1, 2, 3, 4])
print(a+5)
```

```python
# 10.   Make an array of the even numbers from 2 to 20.
np.arange(2,21,2)
```

```python
# 11.  Predict the output: a = np.arange(6) then print(a[2:5])
a=np.arange(6)
print(a) #prints a 6 element array
print(a[2:5]) #prints elements from
```

```python
# 12.  Sort the array [9, 3, 7, 1, 5] from smallest to biggest.
a=np.array([9, 3, 7, 1, 5])
print(np.sort(a))
```

```python
# 13.  Make an array of 12 numbers (0 to 11) and change its shape to 3 rows and 4 columns.
np.arange(0,12).reshape(3,4)
```

```python
# 14.  Find the square root of every number in [1, 4, 9, 16, 25].
a=np.array([1, 4, 9, 16, 25])
print(np.sqrt(a))
```

```python
# 15.  From the array [10, 20, 30, 40, 50, 60], print every second number.
a=np.array([10, 20, 30, 40, 50, 60])
print(a[[1,3,5]])
```

```python
# 16.  Make this array: [10, 20, 30, 40, 50]. Print its shape, its size and its data type.
a=np.arange(10,51,10)
print(a)
print(np.shape(a))
print(np.size(a))
print(a.dtype)
```

```python
# 17.  Count how many numbers in [5, 12, 8, 20, 3, 15] are bigger than 10.
a=np.array([5, 12, 8, 20, 3, 15])
print(sum(np.where(a>10,1,0)))
```

```python
# 18.  Make an array of 5 numbers that are evenly spaced between 0 and 1.
a=np.linspace(0,5,5)
print(a)
```

```python
# 19.  Print the array [1, 2, 3, 4, 5] in reverse order.
a=np.array([1, 2, 3, 4, 5])
print(a[::-1])
```

```python
# 20.  Multiply the arrays [1, 2, 3] and [4, 5, 6], number by number. Print the answer.
a=np.array([1, 2, 3])
b=np.array([4, 5, 6])
print(a*b)
```

```python
# 21.  Predict the output: a = np.array([[1, 2], [3, 4]]) then print(a.shape, a.size)
# output :- (2, 2) 4
```

```python
# 22.  Make a 2 x 3 array with the numbers 1 to 6. Print the number in the second row and third column.
a=np.arange(1,7).reshape(2,3)
print(a)
print(a[1:,2:])
```

```python
# 23.  Change every number smaller than 0 to 0 in the array [-5, 3, -2, 8, -1].
a=np.array([-5, 3, -2, 8, -1])
print(a)
a=np.where(a<0,0,a)
print(a)
```

```python
# 24.  Make an array of 5 random whole numbers between 1 and 100.
np.random.randint(1,100,5)
```

```python
# 25.  For the array [[1, 2, 3], [4, 5, 6]], print only the first row, and then only the first column.
a=np.array([[1, 2, 3], [4, 5, 6]])
print(a[:1,:])
print(a[:,:1])
```

```python
# 26.  Find the position (index) of the biggest number in [14, 38, 22, 45, 9].
a=np.array([14, 38, 22, 45, 9])
print(np.argmax(a)) #argmax() => argument of the maximum, returns the index of the maximum value
```

```python
# 27.  Join the arrays [1, 2, 3] and [4, 5, 6] into one long array.
a=np.array([1, 2, 3])
b=np.array([4, 5, 6])
print(np.hstack((a,b)))
```

```python
# 28.  Predict the output: print(np.array([1, 2, 3]) + np.array([10, 20, 30]))
# output :- [11 22 33]
```

```python
# 29.  Print the array [[1, 2, 3], [4, 5, 6]] with its rows and columns swapped.
a=np.array([[1, 2, 3], [4, 5, 6]])
print(a)
print(a.T)
```

```python
# 30.  Change the Python list [1, 2, 3, 4] into a NumPy array and print its type.
a=[1, 2, 3, 4]
print(a)
print(type(a))

print(a)
a=np.array(a)
print(type(a))
```
