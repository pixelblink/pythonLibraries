```python
# Q.11 Create a 10x10 array with random values and find the minimum and maximum values

arr=np.random.randint(0,1001,100).reshape(10,-1)
print(arr)
max=arr[[0:0,0:0]]
min=arr[[0:0,0:0]]

# min=0
for i in np.nditer(arr):
    if i>max:
        max=i
    elif i<min:
        min=i
print(max)
print(min)
```
