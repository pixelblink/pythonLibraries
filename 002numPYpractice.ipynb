```python
# Q6. Write code to create a 1D array of the first 10 even numbers.

a=np.arange(0,20,2)
print(a)
```

```python
# Q7. Given arr = np.array([1, 2, 3, 4, 5]), write code to print only the last 2 elements using slicing.

arr = np.array([1, 2, 3, 4, 5])
print(arr[3:])
```

```python
# Q8. Write code to create a (3,3) identity matrix using np.eye().

e=np.eye(3,dtype=int)
print(e)
```

```python
# Q9. Given arr = np.array([12, 45, 7, 89, 23]), write code to sort it in ascending order.

arr = np.array([12, 45, 7, 89, 23])
print(np.sort(arr))
```

```python
# Q10. Write code to create a (2,3) array then print its transpose.

a=np.arange(6).reshape(2,3)
print(a)
```

```python
# Q11. Write code to create a (4,4) array using np.arange(16).reshape(4,4), then extract its diagonal elements and compute their sum.

a=np.arange(16).reshape(4,4)
d=np.diag(a)
print(a)
print(d)
print(sum(d))
```

```python
# Q12. Given marks = np.array([45,78,92,34,88,67,55]), write code to find how many students passed (>=40) and the topper's score.

marks = np.array([45,78,92,34,88,67,55])
print(marks[marks>=40])
```

```python
# Q13. Write code to create two (2,2) matrices and compute both their element-wise product and their true matrix product, printing both.

a=np.array([[2,4],[6,8]])
b=np.array([[3,5],[7,9]])

print(a*b)
print(a@b)
```

```python
# Q14. Write code to generate (3,3) array and swap row 0 and 1 and then print

np.random.seed(4)
t=np.random.randint(9,18,(3,3))
print(t)
t[[0,1]]=t[[1,0]]
print(t)
```

```python
# Q15. Write a program to simulate a (5,7) weekly attendance array (1=present, 0=absent) using random values, then print each
# student's total attendance using axis-based aggregation.

np.random.seed(2)
a=np.random.randint(0,2,(5,7))
print(a)
t=np.sum(a,axis=1)
print("student 1 total attendance -",t[0])
print("student 2 total attendance -",t[1])
print("student 3 total attendance -",t[2])
print("student 4 total attendance -",t[3])
print("student 5 total attendance -",t[4])
```

```python
# Q16. Create a NumPy array of numbers from 1 to 10. Write code to:
# 1. Square each element.

n=np.arange(1,11)
print(n)
print(n**2)
```

```python
# Q17. Create a 5×5 array filled with random integers between 1 and 50. Write code to:
# 1. Find all even numbers in the array.
# 2. Replace them with 0.

np.random.seed(4)

a=np.random.randint(1,51,(5,5))
print(np.where(a%2==0,0,a))
```
